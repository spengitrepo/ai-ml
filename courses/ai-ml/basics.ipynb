```
transactions = [
    {"customer": "A", "amount": 1200, "status": "completed"},
    {"customer": "B", "amount": 500, "status": "completed"},
    {"customer": "C", "amount": 3000, "status": "failed"},
    {"customer": "D", "amount": 800, "status": "completed"},
]


#total = sum(t["amount"] for t in transactions)

x_total=0
for  t in transactions:
  x_total=x_total+t["amount"]
print(x_total)


for  t in transactions:
  if t['status'] == "completed":
    print(t)

def calc_sucess_rate(transactions):
  success= sum(
      1 for t in transactions
      if t['status'] == "completed"
  )
  return success/len(transactions)

rate = calc_sucess_rate(transactions)
print(f"Success Rate: {rate:.2%}")
```
